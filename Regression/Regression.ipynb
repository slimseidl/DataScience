```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

df = pd.read_csv('house_prices.csv')
df.head()
```

```python
# Have to add intercept col 
df['intercept'] = 1
```

```python
# Least Squares Method
lm = sm.OLS(df['price'], df[['intercept', 'area']])
```

```python
# Fitting a line using least squares
results = lm.fit()
```

```python
# Summary of regression model
results.summary()
```

### Interpreting the Results 
- Intercept = Intercept Coef b0
- Slope = Area coef b1
- y = intercept + slope (b0 + b1x)
- Predicted price = 9588 + 348x
- x = area of house 
- P Value - P>|t|
    - P value closer to 0 = statistically significant
    - Area is significant to predicting price
- R squared  value (top right) - closer to 1, better the model fits, closer to 0 = worse fit
- R squared is the square of correl coeff
- 67% of the price can be explained by house area

#### Logistic Regression

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm


df = pd.read_csv('fraud.csv')
df.head()
```

```python
df[["weekend","weekday"]] = pd.get_dummies(df["day"]).astype(int)
```

```python
df[["not_fraud","fraud"]] = pd.get_dummies(df["fraud"]).astype(int)
df.drop("not_fraud",inplace=True, axis=1)
```

```python
df.drop(columns=("weekend"),inplace=True)
```

```python
df.head()
```

```python
# Proportion of raud and proportion of weekday
print(df["fraud"].mean())
print(df["weekday"].mean())
```

```python
df.query("fraud == 1")['duration'].mean()
```

```python
df["intercept"] = 1

model = sm.Logit(df["fraud"],df[["intercept","duration","weekday"]])
```

```python
results=model.fit()
```

```python
results.summary2()
```

```python
results.summary()
```

```python
# to exponentiate
np.exp(-1.4637), np.exp(-2.5465)
```

### Interpret Results of Logistic Regression

- Exponentiate the coefficients
    - -1.4637 becomes .231
    - -2.5465 becomes 12.762 for weekday
    - np.exp(value)
    - For each 1 unit increase in duration, fraud is 0.23 times as likely, holding all other variables (weekday) constant.
    - Fraud is 12.76 times as likely on weekdays than weekends, holding all other variables (duration) constant.
